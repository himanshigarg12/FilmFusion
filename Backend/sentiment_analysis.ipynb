```python
# Cell 1 - Import all libraries we need
# These are like "tools" we need for ML

import pandas as pd          # for reading and cleaning data
import numpy as np           # for numbers and maths
import pickle                # for saving our trained model
import os                    # for file operations

# ML tools from scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("✅ All libraries imported successfully!")
```

```python
# Cell 2 - Load and explore the dataset
# We are reading our CSV file using Pandas

# Load the dataset
df = pd.read_csv('IMDB Dataset.csv')

# Show first 5 rows to see what data looks like
print("First 5 rows of dataset:")
print(df.head())

print("\n")

# Show shape - how many rows and columns
print(f"Total reviews: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")

print("\n")

# Show how many positive and negative reviews
print("Sentiment distribution:")
print(df['sentiment'].value_counts())

print("\n✅ Dataset loaded successfully!")
```

```python
# Cell 3 - Clean the data
# Remove HTML tags, special characters etc.

import re

def clean_text(text):
    # Remove HTML tags like <br />, <b> etc.
    text = re.sub(r'<.*?>', '', text)
    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # Convert to lowercase
    text = text.lower()
    # Remove extra spaces
    text = text.strip()
    return text

# Apply cleaning to all reviews
df['clean_review'] = df['review'].apply(clean_text)

# Convert sentiment to numbers
# positive = 1, negative = 0
df['label'] = df['sentiment'].map({'positive': 1, 'negative': 0})

# Show cleaned data
print("Before cleaning:")
print(df['review'][0][:100])
print("\nAfter cleaning:")
print(df['clean_review'][0][:100])
print("\nLabel (1=positive, 0=negative):", df['label'][0])
print("\n✅ Data cleaned successfully!")
```

```python
# Cell 4 - Split data and train the model
# This is where actual ML magic happens!

# Separate reviews and labels
X = df['clean_review']  # input (reviews)
y = df['label']         # output (0 or 1)

# Split into training and testing data
# 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2,      # 20% for testing
    random_state=42     # same split every time
)

print(f"Training reviews: {len(X_train)}")
print(f"Testing reviews: {len(X_test)}")

# Convert text to numbers using TF-IDF
# TF-IDF = how important each word is in a review
print("\nConverting text to numbers...")
vectorizer = TfidfVectorizer(
    max_features=10000,  # use top 10000 words
    stop_words='english' # ignore common words like "the", "is"
)

# Learn vocabulary from training data
X_train_tfidf = vectorizer.fit_transform(X_train)

# Apply same vocabulary to test data
X_test_tfidf = vectorizer.transform(X_test)

print(f"Vocabulary size: {len(vectorizer.vocabulary_)}")
print("\n✅ Data prepared for training!")
```

```python
# Cell 5 - Train the ML model
# This is where computer LEARNS from data!

print("Training Naive Bayes model...")
print("Please wait...")

# Create and train Naive Bayes model
nb_model = MultinomialNB()
nb_model.fit(X_train_tfidf, y_train)

# Test the model
nb_predictions = nb_model.predict(X_test_tfidf)
nb_accuracy = accuracy_score(y_test, nb_predictions)

print(f"\n✅ Naive Bayes Training Complete!")
print(f"Accuracy: {nb_accuracy * 100:.2f}%")

print("\n" + "="*50)
print("Training Logistic Regression model...")

# Create and train Logistic Regression model
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train_tfidf, y_train)

# Test the model
lr_predictions = lr_model.predict(X_test_tfidf)
lr_accuracy = accuracy_score(y_test, lr_predictions)

print(f"\n✅ Logistic Regression Training Complete!")
print(f"Accuracy: {lr_accuracy * 100:.2f}%")

print("\n" + "="*50)
print("COMPARISON:")
print(f"Naive Bayes Accuracy:       {nb_accuracy * 100:.2f}%")
print(f"Logistic Regression Accuracy: {lr_accuracy * 100:.2f}%")

if lr_accuracy > nb_accuracy:
    print("\n🏆 Winner: Logistic Regression!")
    best_model = lr_model
else:
    print("\n🏆 Winner: Naive Bayes!")
    best_model = nb_model

print("\n✅ Both models trained successfully!")
```

```python
# Cell 6 - Evaluate the model properly
# This covers your syllabus Chapter 19-21!

print("=== CONFUSION MATRIX ===")
cm = confusion_matrix(y_test, lr_predictions)
print(cm)

print("\n=== DETAILED REPORT ===")
print(classification_report(y_test, lr_predictions, 
      target_names=['Negative', 'Positive']))

print("\n=== WHAT THIS MEANS ===")
print(f"True Positives  (correctly predicted positive): {cm[1][1]}")
print(f"True Negatives  (correctly predicted negative): {cm[0][0]}")
print(f"False Positives (predicted positive but wrong): {cm[0][1]}")
print(f"False Negatives (predicted negative but wrong): {cm[1][0]}")

print("\n✅ Model evaluation complete!")
```

```python
# Cell 7 - Save the trained model
# Save as .pkl file so Flask can use it!

import os

# Create models folder path
models_dir = os.path.join(os.getcwd(), 'models')
os.makedirs(models_dir, exist_ok=True)

# Save Logistic Regression model (winner!)
model_path = os.path.join(models_dir, 'sentiment_model.pkl')
with open(model_path, 'wb') as f:
    pickle.dump(lr_model, f)
print(f"✅ Model saved: {model_path}")

# Save TF-IDF Vectorizer (needed for predictions!)
vectorizer_path = os.path.join(models_dir, 'vectorizer.pkl')
with open(vectorizer_path, 'wb') as f:
    pickle.dump(vectorizer, f)
print(f"✅ Vectorizer saved: {vectorizer_path}")

# Test the saved model works
print("\n=== TESTING SAVED MODEL ===")

# Load model back
with open(model_path, 'rb') as f:
    loaded_model = pickle.load(f)

with open(vectorizer_path, 'rb') as f:
    loaded_vectorizer = pickle.load(f)

# Test with sample reviews
test_reviews = [
    "This movie was absolutely amazing! Best film ever!",
    "Terrible movie, complete waste of time and money",
    "Shah Rukh Khan was brilliant in this role",
    "Boring and disappointing, very bad acting"
]

print("\nTesting with sample reviews:")
for review in test_reviews:
    cleaned = clean_text(review)
    vectorized = loaded_vectorizer.transform([cleaned])
    prediction = loaded_model.predict(vectorized)[0]
    probability = loaded_model.predict_proba(vectorized)[0]
    sentiment = "POSITIVE 😍" if prediction == 1 else "NEGATIVE 😤"
    confidence = max(probability) * 100
    print(f"\nReview: {review[:50]}...")
    print(f"Prediction: {sentiment} ({confidence:.1f}% confident)")

print("\n✅ Model saved and tested successfully!")
print("🎉 Your AI Sentiment Model is ready!")
```
